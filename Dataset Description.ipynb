```python
def generate_comparison_table(datasets_info):
    """
    datasets_info: List of tuples [(dataframe, sequences, "Name"), ...]
    """
    all_stats = []
    
    for data, sequences, name in datasets_info:
        samples_per_patient = data.groupby('id').size()
        y_labels = [s['Y'] for s in sequences]
        
        # Calculate TIR metrics per patient
        patient_metrics = data.groupby('id')['gl'].apply(lambda x: pd.Series({
            'tb54': (x < 54).mean() * 100,
            'tb70': (x < 70).mean() * 100,
            'tir': ((x >= 70) & (x <= 180)).mean() * 100,
            'ta180': (x > 180).mean() * 100,
            'ta250': (x > 250).mean() * 100
        })).unstack()

        stats = {
            "Metric": name,
            "Number of patients": data['id'].nunique(),
            "Number of samples per patient (min-max)": f"{samples_per_patient.min()} - {samples_per_patient.max()}",
            "Number of samples per patient (mean ± SD)": f"{samples_per_patient.mean():.2f} (±{samples_per_patient.std():.2f})",
            "Total sequences": len(sequences),
            "Positive labelled (Y=1)": sum(y_labels),
            "Negative labelled (Y=0)": len(y_labels) - sum(y_labels),
            "Glucose min-max": f"{data['gl'].min()} - {data['gl'].max()}",
            "Glucose mean (SD)": f"{data['gl'].mean():.2f} (±{data['gl'].std():.2f})",
            "Time Below 54 mg/dL (%)": f"{patient_metrics['tb54'].mean():.2f}% (±{patient_metrics['tb54'].std():.2f})",
            "Time Below 70 mg/dL (%)": f"{patient_metrics['tb70'].mean():.2f}% (±{patient_metrics['tb70'].std():.2f})",
            "Time In Range 70-180 mg/dL (%)": f"{patient_metrics['tir'].mean():.2f}% (±{patient_metrics['tir'].std():.2f})",
            "Time Above 180 mg/dL (%)": f"{patient_metrics['ta180'].mean():.2f}% (±{patient_metrics['ta180'].std():.2f})",
            "Time Above 250 mg/dL (%)": f"{patient_metrics['ta250'].mean():.2f}% (±{patient_metrics['ta250'].std():.2f})"

            #agregar cuantas lecturas tienen las secuencias (promedio, minimo, maximo, deviacion estandar)
            #agregar numero de secuencias inciales y el numero de secuencias eliminadas
        }
        all_stats.append(stats)
    
    # Transpose for the final look
    return pd.DataFrame(all_stats).set_index("Metric").T
```

```python
generate_comparison_table([(Replace, sequences_Replace, "Replace Dataset")])
```

```python
generate_comparison_table([(Replace, night_sequences_Replace, "Replace Dataset (Night)")])
```
