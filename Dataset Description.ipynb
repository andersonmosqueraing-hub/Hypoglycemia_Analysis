```python
import pandas as pd
import numpy as np

def describe_dataset(data, sequences):
    # --- 1. Estadísticas de Pacientes y Muestras ---
    samples_per_patient = data.groupby('id')['gl'].count()
    
    stats = {
        "Number of patients": data['id'].nunique(),
        "Number of samples per patient (min-max)": f"{samples_per_patient.min()} - {samples_per_patient.max()}",
        "Number of samples per patient (mean ± SD)": f"{samples_per_patient.mean():.2f} (±{samples_per_patient.std():.2f})",
    }

    # --- 2. Estadísticas de Secuencias ---
    y_labels = [s['Y'] for s in sequences]
    stats.update({
        "Total sequences": len(sequences),
        "Positive labelled (Y=1)": sum(y_labels),
        "Negative labelled (Y=0)": len(y_labels) - sum(y_labels)
    })

    # --- 3. Estadísticas de Glucosa (Global) ---
    stats.update({
        "Glucose min-max": f"{data['gl'].min()} - {data['gl'].max()}",
        "Glucose mean (SD)": f"{data['gl'].mean():.2f} (±{data['gl'].std():.2f})"
    })

    # --- 4. Métricas de Tiempo en Rango (TIR) por Paciente ---
    # Calculamos los porcentajes para cada paciente individualmente primero
    def get_tir_metrics(df):
        total = len(df)
        return pd.Series({
            'tb54': (df['gl'] < 54).sum() / total * 100,
            'tb70': (df['gl'] < 70).sum() / total * 100,
            'tir': ((df['gl'] >= 70) & (df['gl'] <= 180)).sum() / total * 100,
            'ta180': (df['gl'] > 180).sum() / total * 100,
            'ta250': (df['gl'] > 250).sum() / total * 100
        })

    patient_metrics = data.groupby('id').apply(get_tir_metrics)

    # Agregamos las métricas de tiempo al diccionario final
    ranges = [
        ('tb54', '% Time below 54 mg/dL'),
        ('tb70', '% Time below 70 mg/dL'),
        ('tir', '% Time in range 70-180 mg/dL'),
        ('ta180', '% Time above 180 mg/dL'),
        ('ta250', '% Time above 250 mg/dL')
    ]

    for col, label in ranges:
        stats[f"{label} (min-max)"] = f"{patient_metrics[col].min():.2f}% - {patient_metrics[col].max():.2f}%"
        stats[f"{label} (mean ± SD)"] = f"{patient_metrics[col].mean():.2f}% (±{patient_metrics[col].std():.2f}%)"

    return pd.Series(stats)

# Ejecución:
# results = describe_dataset(data, sequences)
# print(results.to_string())
```
